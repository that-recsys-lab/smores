```python
from modules.classes import Document, Provider, Consumer, Recommender
from modules.samplers import provider_sampler, consumer_sampler
from modules.genre_preferences import generate_genre_preferences
from modules.recommenders import PopularRecommender

# Initialize recommender
popular_recommender = PopularRecommender(recommender_id="popular")

# Simulate cold-start users
new_users = [
    Consumer(consumer_id=1, sensitivity=0.5, category_preferences={}),
    Consumer(consumer_id=2, sensitivity=0.6, category_preferences={}),
]
recommendations = popular_recommender.recommend_documents(new_users, slate_size=3)

# Validate results
for user_id, docs in recommendations.items():
    assert len(docs) > 0, f"Consumer {user_id} received no recommendations"
    assert all(doc.doc_id in [item["movieId"] for item in popular_recommender.popular_items] for doc in docs), \
        f"Consumer {user_id} received invalid recommendations"

print("All tests passed!")
```
