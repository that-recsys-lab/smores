```python
import pandas as pd

ratings = pd.read_csv('ratings.csv', dtype={'movieId': 'str'})  # userId, movieId, rating, timestamp
movies = pd.read_csv('movies.csv', dtype={'movieId': 'str'})  # movieId, title, genres
links = pd.read_csv('links.csv', dtype={'movieId': 'str', 'tmdbId': 'str', 'imdbId': 'str'})  # movieId, imdbId, tmdbId
movies_metadata = pd.read_csv('movies_metadata.csv', dtype={'id': 'str'}, low_memory=False)  # detailed metadata

# Merge ratings with movies to add titles and genres
ratings_movies = ratings.merge(movies, on='movieId', how='left')

# Merge links with movies_metadata to add detailed metadata
links_metadata = links.merge(movies_metadata, left_on='tmdbId', right_on='id', how='left')

# merge the detailed metadata with ratings and movies data
full_dataset = ratings_movies.merge(links_metadata, on='movieId', how='left')

# only relevant columns for ratings
relevant_columns = ['userId', 'movieId', 'rating', 'timestamp']
cleaned_dataset = full_dataset[relevant_columns]
cleaned_dataset.to_csv('merge_movie_ratings.csv', index=False)
```
