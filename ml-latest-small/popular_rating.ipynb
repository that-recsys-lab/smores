```python
import pandas as pd

file_path = 'ratings_with_providers.csv'  
ratings_data = pd.read_csv(file_path)

def calculate_popularity(ratings):
    popularity = (
        ratings.groupby(['movieId', 'title'])  
        .size()  
        .reset_index(name='rating_count') 
        .sort_values(by='rating_count', ascending=False)  
    )
    return popularity

top_n = 10  
popular_items = calculate_popularity(ratings_data)
top_n_popular_items = popular_items.head(top_n)

print(top_n_popular_items)
```
